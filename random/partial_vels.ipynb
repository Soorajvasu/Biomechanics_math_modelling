```python
import sympy as sp
import sympy.physics.mechanics as me  
me.init_vprinting()
```

```python
A, B, E = sp.symbols('A B E', cls=me.ReferenceFrame)
q1, q2, q3 = me.dynamicsymbols('q1:4')
u1, u2, u3 = me.dynamicsymbols('u1:4')
theta = me.dynamicsymbols('theta')
t = me.dynamicsymbols._t

P1 = me.Point('P1')
O = me.Point('O')

u1 = q1.diff(t)
u2 = q2.diff(t)
u3 = q3.diff(t)
u1, u2, u3
```

```python
B.orient_axis(A, theta, A.y)
E.orient_axis(B, q3, B.z)

P1.set_pos(O, q1*B.x + q2*B.y)
```

```python
B.ang_vel_in(A).express(B)
```

```python
O.set_vel(A, 0)
P1.set_vel(B, P1.pos_from(O).dt(B))
P1.v1pt_theory(O, A, B)
```

```python
P1_V_A = P1.pos_from(O).dt(B) + me.cross(B.ang_vel_in(A), P1.pos_from(O))
P1_V_A.express(E)
```

```python
P1_V_A.express(A)
```

```python
P1.set_vel(A, P1_V_A)
```

```python
P1_V_A.diff((q1).diff(t), A), P1_V_A.diff((q2).diff(t), A), P1_V_A.diff((q3).diff(t), A)
```

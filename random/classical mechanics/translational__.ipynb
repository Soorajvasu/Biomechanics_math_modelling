```python
import sympy as sp
import sympy.physics.mechanics as me
me.init_vprinting()
```

```python
N, A, B = sp.symbols('N A B', cls=me.ReferenceFrame)
alpha, beta = me.dynamicsymbols('alpha beta', real="True")
h, d, w, c , l = sp.symbols('h d w c l', real="True", positive="True")
s = me.dynamicsymbols('s')
t = me.dynamicsymbols._t
```

```python
A.orient_axis(N, alpha, N.z)
B.orient_axis(A, beta, A.x)
```

```python
A.ang_vel_in(N)
```

```python
B.ang_vel_in(N)
```

```python
r_O_P = h*N.z
r_P_S = -d*A.x
r_S_Q = -w*B.x - (c + l/2)*B.z

r_O_P, r_P_S, r_S_Q
```

```python
V_s_N = r_P_S.dt(A) + me.cross(A.ang_vel_in(N), r_P_S)
V_s_N
```

```python
V_s_N = r_P_S.dt(A) + me.cross(A.ang_vel_in(N), r_P_S)
V_s_N
```

```python
O = me.Point('O')
P = me.Point('P')
S = me.Point('S')
Q = me.Point('Q')
R = me.Point('R')

P.set_pos(O, h*N.z)
S.set_pos(P, -d*A.x)
Q.set_pos(S, -w*B.x - (c + l/2)*B.z)
R.set_pos(Q, l*B.z + s*B.x)
```

```python
S.set_vel(N, V_s_N)

Q.v2pt_theory(S, N, B)
```

```python
Q.set_vel(B, r_S_Q.dt(B))
Q.v1pt_theory(S, N, B)
```

```python
B_v_R = s.diff(t)*B.x
B_v_R
```

```python
r_S_R = R.pos_from(S)
me.cross(B.ang_acc_in(A), r_S_R)
```

```python
me.cross(B.ang_vel_in(A),me.cross(B.ang_vel_in(A),r_S_Q))
```

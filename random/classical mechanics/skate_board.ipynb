```python
import sympy as sp
import sympy.physics.mechanics as me
me.init_vprinting()

N, A = sp.symbols('N A', cls=me.ReferenceFrame)
theta = me.dynamicsymbols('theta', real="True")
x, y = me.dynamicsymbols('x y')
t = me.dynamicsymbols._t
```

```python
A.orient_axis(N, theta, N.z)

O = me.Point('O')
P = me.Point('P')

P.set_pos(O, x*N.x + y*N.y)
```

```python
P.pos_from(O).dt(N)
```

```python
P.pos_from(O).dt(A) + me.cross(A.ang_vel_in(N),P.pos_from(O))
```

```python
P.pos_from(O).dt(A)
```

```python
me.cross(A.ang_vel_in(N),P.pos_from(O))
```

```python
import sympy as sp
import sympy.physics.mechanics as me 
me.init_vprinting()
```

```python
N = sp.symbols('N', cls=me.ReferenceFrame)
m1, m2, m3, m4 = sp.symbols('m1:5', real="True", positive="True")
x1, x2, x3, x4 = sp.symbols('x1:5', real="True", positive="True")
y1, y2, y3, y4 = sp.symbols('y1:5', real="True", positive="True")
z1, z2, z3, z4 = sp.symbols('z1:5', real="True", positive="True")

r1 = x1*N.x + y1*N.y + z1*N.z
r2 = x2*N.x + y2*N.y + z2*N.z
r3 = x3*N.x + y3*N.y + z3*N.z
r4 = x4*N.x + y4*N.y + z4*N.z

zeroth_moment = m1 + m2 + m3 + m4

first_moment = m1*r1 + m2*r2 + m3*r3 + m4*r4

com = first_moment/zeroth_moment
com
```

```python
Ix = (
    m1*me.cross(r1, me.cross(N.x, r1)) + 
    m2*me.cross(r2, me.cross(N.x, r2)) + 
    m3*me.cross(r3, me.cross(N.x, r3)) + 
    m4*me.cross(r4, me.cross(N.x, r4))
)
Ix
```

```python
Iy = (
    m1*me.cross(r1, me.cross(N.y, r1)) + 
    m2*me.cross(r2, me.cross(N.y, r2)) + 
    m3*me.cross(r3, me.cross(N.y, r3)) + 
    m4*me.cross(r4, me.cross(N.y, r4))
)
Iy
```

```python
Iz = (
    m1*me.cross(r1, me.cross(N.z, r1)) + 
    m2*me.cross(r2, me.cross(N.z, r2)) + 
    m3*me.cross(r3, me.cross(N.z, r3)) + 
    m4*me.cross(r4, me.cross(N.z, r4))
)
Iz
```

```python
Ix_num = Ix.xreplace({x1: 3, x2: 0, x3: 4, x4: 0,
             y1: 4, y2: 3, y3: 0, y4: 4,
             z1: 0, z2: 4, z3: 3, z4: 3,}).simplify()
Iy_num = Iy.xreplace({x1: 3, x2: 0, x3: 4, x4: 0,
             y1: 4, y2: 3, y3: 0, y4: 4,
             z1: 0, z2: 4, z3: 3, z4: 3,}).simplify()
Iz_num = Iz.xreplace({x1: 3, x2: 0, x3: 4, x4: 0,
             y1: 4, y2: 3, y3: 0, y4: 4,
             z1: 0, z2: 4, z3: 3, z4: 3,}).simplify()

Ix_num, Iy_num, Iz_num
```

```python
sp.pprint(f'Ixy: {Ix_num.dot(N.y)}')
sp.pprint(f'Iyx: {Iy_num.dot(N.x)}')
```

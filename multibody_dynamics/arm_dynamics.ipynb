```python
import numpy as np
import sympy as sp
import sympy.physics.mechanics as me
me.init_vprinting()
```

```python
t = me.dynamicsymbols._t

N, A, B, C, D, E, F, G = sp.symbols('N A B C D E F G', cls=me.ReferenceFrame)
O, L1, L2, L3, L4, L5, L6, L7 = sp.symbols('O L_1:8', cls=me.Point)

# Generalised coordinated and speeds
q1, q2, q3, q4, q5, q6, q7 = me.dynamicsymbols('q1:8')
u1, u2, u3, u4, u5, u6, u7 = me.dynamicsymbols('u1:8')

# lengths, masses
l1 ,l2, l3, l4, l5, l6, l7 = sp.symbols('l1:8', positive="True", real="True")
m1 ,m2, m3, m4, m5, m6, m7, g = sp.symbols('m1:8 g', positive="True", real="True")

# Inertias about CoM
I_l1 = me.inertia(A, 1, 1, 1, 1, 1, 1)
I_l2 = me.inertia(B, 1, 1, 1, 1, 1, 1)
I_l3 = me.inertia(C, 1, 1, 1, 1, 1, 1)
I_l4 = me.inertia(D, 1, 1, 1, 1, 1, 1)
I_l5 = me.inertia(E, 1, 1, 1, 1, 1, 1)
I_l6 = me.inertia(F, 1, 1, 1, 1, 1, 1)
I_l7 = me.inertia(G, 1, 1, 1, 1, 1, 1)

# Joint torques
T1, T2, T3, T4, T5, T6, T7 = me.dynamicsymbols('T_joint1:8')
```

```python

q = sp.Matrix([q1, q2, q3, q4, q5, q6, q7])
qdot = q.diff(t)
u = sp.Matrix([u1, u2, u3, u4, u5, u6, u7])

qdot_repl = dict(zip(qdot,u))
qdot_repl
```

```python
# Frame Orientation
A.orient_axis(N, q1, N.y)
B.orient_axis(A, q2, A.x)
C.orient_axis(B, q3, B.y)
D.orient_axis(C, q4, C.x)
E.orient_axis(D, q5, D.y)
F.orient_axis(E, q6, E.x)
G.orient_axis(F, q7, F.y)

# CoM
L1.set_pos(O, l1/2*A.y)
L2.set_pos(L1, (l1/2+l2/2)*B.y)
L3.set_pos(L2, (l2/2+l3/2)*C.y)
L4.set_pos(L3, (l3/2+l4/2)*D.y)
L5.set_pos(L4, (l4/2+l5/2)*E.y)
L6.set_pos(L5, (l5/2+l6/2)*F.y)
L7.set_pos(L6, (l6/2+l7/2)*G.y)
```

```python
# Position check

L7.pos_from(O)
```

```python
# velocities and acceleration

A.set_ang_vel(N, u1*N.y)
B.set_ang_vel(N, u1*N.y + u2*A.x)
C.set_ang_vel(N, u1*N.y + u2*A.x + u3*B.y)
D.set_ang_vel(N, u1*N.y + u2*A.x + u3*B.y + u4*C.x)
E.set_ang_vel(N, u1*N.y + u2*A.x + u3*B.y + u4*C.x + u5*D.y)
F.set_ang_vel(N, u1*N.y + u2*A.x + u3*B.y + u4*C.x + u5*D.y + u6*E.x)
G.set_ang_vel(N, u1*N.y + u2*A.x + u3*B.y + u4*C.x + u5*D.y + u6*E.x + u7*F.y)

me.find_dynamicsymbols(G.ang_vel_in(N), reference_frame=N), me.find_dynamicsymbols(G.ang_acc_in(N), reference_frame=N)
```

```python
O.set_vel(N, 0)

L1.v2pt_theory(O, N, A)
L2.v2pt_theory(L1, N, B)
L3.v2pt_theory(L2, N, C)
L4.v2pt_theory(L3, N, D)
L5.v2pt_theory(L4, N, E)
L6.v2pt_theory(L5, N, F)
L7.v2pt_theory(L6, N, G)

L1.set_acc(N, L1.acc(N).xreplace(qdot_repl))
L2.set_acc(N, L2.acc(N).xreplace(qdot_repl))
L3.set_acc(N, L3.acc(N).xreplace(qdot_repl))
L4.set_acc(N, L4.acc(N).xreplace(qdot_repl))
L5.set_acc(N, L5.acc(N).xreplace(qdot_repl))
L6.set_acc(N, L6.acc(N).xreplace(qdot_repl))
L7.set_acc(N, L7.acc(N).xreplace(qdot_repl))

me.find_dynamicsymbols(L7.vel(N), reference_frame=N), me.find_dynamicsymbols(L7.acc(N), reference_frame=N)
```

```python
# Forces and torques on the body

R_l1 = -m1*g*N.z
R_l2 = -m2*g*N.z
R_l3 = -m3*g*N.z
R_l4 = -m4*g*N.z
R_l5 = -m5*g*N.z
R_l6 = -m6*g*N.z
R_l7 = -m7*g*N.z

T_A = T1*A.y + T2*B.x
T_B = T2*B.x + T3*C.y
T_C = T3*C.y + T4*D.x
T_D = T4*D.x + T5*E.y
T_E = T5*E.y + T6*F.x
T_F = T6*F.x + T7*G.y
T_G = T7*G.y
```

```python
# Generalised active and inertia forces

points = [L1, L2, L3, L4, L5, L6, L7]
forces = [R_l1, R_l2, R_l3, R_l4, R_l5, R_l6, R_l7]
masses = [m1 ,m2, m3, m4, m5, m6, m7]

frames = [A, B, C, D, E, F, G]
torques = [T_A, T_B, T_C, T_D, T_E, T_F, T_G]
inertias = [I_l1, I_l2, I_l3, I_l4, I_l5, I_l6, I_l7]

Fr_bar = []
Frs_bar = []

for ur in [u1, u2, u3, u4, u5, u6, u7]:

    Fr = 0
    Frs = 0

    for Pi, Ri, mi in zip(points, forces, masses):
        vr = Pi.vel(N).diff(ur, N)
        Fr += vr.dot(Ri)
        Rs = -mi*Pi.acc(N)
        Frs += vr.dot(Rs)

    for Bi, Ti, Ii in zip(frames, torques, inertias):
        wr = Bi.ang_vel_in(N).diff(ur, N)
        Fr += wr.dot(Ti)
        Ts = -(Bi.ang_acc_in(N).dot(Ii) +
               me.cross(Bi.ang_vel_in(N), Ii).dot(Bi.ang_vel_in(N)))
        Frs += wr.dot(Ts)

    Fr_bar.append(Fr)
    Frs_bar.append(Frs)

Fr = sp.Matrix(Fr_bar)
Frs = sp.Matrix(Frs_bar)
```

### Equations of motion

```python
EOM = Fr + Frs
# EOM = sp.trigsimp(EOM.simplify(), method='fu')
# EOM = EOM.simplify()
```

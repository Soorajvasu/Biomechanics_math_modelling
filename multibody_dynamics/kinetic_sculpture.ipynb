```python
import sympy as sp
import sympy.physics.mechanics as me
me.init_vprinting()

N, A, B = sp.symbols('N A B', cls=me.ReferenceFrame)
alpha, beta = me.dynamicsymbols('alpha beta', real="True")
d, w, c , l = sp.symbols('d w c l', real="True", positive="True")
h, s = me.dynamicsymbols('h s')
t = me.dynamicsymbols._t
```

```python
A.orient_axis(N, alpha, N.z)
B.orient_axis(A, beta, A.x)

O = me.Point('O')
P = me.Point('P')
S = me.Point('S')
Q = me.Point('Q')
R = me.Point('R')

P.set_pos(O, h*N.z)
S.set_pos(P, -d*A.x)
Q.set_pos(S, -w*B.x - (c + l/2)*B.z)
R.set_pos(S, (l/2 - c)*B.z + (s-w)*B.x)
```

```python
# velocity of point P

V_p_o = P.pos_from(O).dt(A) + me.cross(A.ang_vel_in(N), P.pos_from(O))
P.set_vel(N, V_p_o)

V_p_o
```

```python
# velocity of point S

V_s_o = S.v2pt_theory(P, N, A)
S.set_vel(N, V_s_o)

V_s_o
```

```python
# velocity of point Q

V_q_o = Q.v2pt_theory(S, N, B)
Q.set_vel(N, V_q_o)

V_q_o
```

```python
# velocity of point R
V_r_B = R.pos_from(S).dt(B)
R.set_vel(B, V_r_B)
V_r_o = R.v1pt_theory(S, N, B)

V_r_o
```

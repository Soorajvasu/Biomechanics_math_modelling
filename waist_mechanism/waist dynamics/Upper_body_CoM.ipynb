```python
import sympy as sp
import sympy.physics.mechanics as me
me.init_vprinting(use_latex='mathjax')
```

```python
m1, m2, m3 = sp.symbols('m1, m2, m3')
x1, x2, x3 = me.dynamicsymbols('x1, x2, x3')
y1, y2, y3 = me.dynamicsymbols('y1, y2, y3')
z1, z2, z3 = me.dynamicsymbols('z1, z2, z3')

A = sp.symbols('A', cls=me.ReferenceFrame)

zeroth_moment = (m1 + m2 + m3)

first_moment = (m1*(x1*A.x + y1*A.y + z1*A.z) +
                m2*(x2*A.x + y2*A.y + z2*A.z) +
                m3*(x3*A.x + y3*A.y + z3*A.z))
```

```python
CoM =  first_moment/zeroth_moment
```

```python
values = {
    m1: 25.0,
    m2: 30.0,
    m3: 10.0,
    x1: 0.37*(0.5)*sp.sin(20*sp.pi/180),
    x2: (0.37*sp.sin(20*sp.pi/180)) + 0.35,
    x3: (0.37*sp.sin(20*sp.pi/180)) + 0.7 + 0.1,
    y1: 0.0,
    y2: 0.0,
    y3: 0.0,
    z1: 0.37*0.5*sp.cos(20*sp.pi/180),
    z2: 0.37*sp.cos(20*sp.pi/180),
    z3: 0.37*sp.cos(20*sp.pi/180),
}

CoM = CoM.xreplace(values)
CoM.evalf(n=10)
```
